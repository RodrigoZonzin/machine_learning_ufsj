```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('data/fetal_health.csv')
mycolor = '#c78883'
```

```python
df.columns
```

```python
## Transformando as classes 1, 2 e 3 em inteiros (em vez de float)

df['fetal_health'] = df['fetal_health'].astype(int)
```

## Separando os conjuntos $X_{train}$, $Y_{train}$ e $X_{test}$, $Y_{test}$

$X = X_{train} \cup X_{test} $ 

$Y = Y_{train} \cup Y_{test} $ 

```python
print("Dimensão dos dados: ", df.shape)
```

```python
from pandas.plotting import scatter_matrix
plt.figure(figsize=(200, 150))
scatter_matrix(df)
plt.savefig('bigpic.png')
```

```python
corr_matrix = df.corr()
```

```python
corr_matrix
```

```python
#separando os dados em dois conjuntos $
train, test = train_test_split(df, test_size=0.1, random_state=42)
```

```python
Xtrain = train.drop('fetal_health', axis=1)
Xtest  = test.drop('fetal_health', axis=1)

Ytrain  = train['fetal_health']
Ytest   = test['fetal_health']


print(f'X train shape: {Xtrain.shape} e X test shape: {Xtest.shape}')
print(f'Y train shape: {Ytrain.shape} e Y test shape: {Ytest.shape}')
```

### Analisando o conjunto $Y_{train}$

```python
plt.figure(figsize=(8, 6))
plt.grid(True, alpha = 0.5)
Ytrain.value_counts().plot(kind='bar', cmap = 'cividis')
plt.ylabel("Número de instâncias")
plt.xlabel("Classes")
plt.xticks(fontsize = 12)
plt.yticks(fontsize = 12)
plt.tight_layout()
plt.savefig('results/imgs/frequencia_classes_Ytrain.png')
plt.show()
```

### Salvando dados

```python
Ytrain.to_pickle('data/ytrain.pkl')
Ytest.to_pickle('data/ytest.pkl')

Xtrain.to_pickle('data/xtrain.pkl')
Xtest.to_pickle('data/xtest.pkl')
```
